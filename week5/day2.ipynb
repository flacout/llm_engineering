# Idea behind RAG

First there is an extra LLM, an Encoding LLM.
Here is the flow:
1. Use Encoding LLM to make vectors for all the text documents in the knowledge-base.
2. Store these vectors in a Vector Datastore.
3. A user ask a question in the user prompt.
4. The Encoding LLM make a vector for this prompt.
5. Search the Vector Datastore for Vectors that are close to this PromptVector.
6. Retreive the result of this query. But the text associated with these Results Vector.
7. Add this text as context to the Prompt.
8. Send this Contextualized Prompt to the Chat LLM
9. Return the answer to the User

```python
# imports for langchain

from langchain.document_loaders import DirectoryLoader, TextLoader
from langchain.text_splitter import CharacterTextSplitter
```

```python
# price is a factor for our company, so we're going to use a low cost model

MODEL = "gpt-4o-mini"
db_name = "vector_db"
```

```python
# Load environment variables in a file called .env

load_dotenv()
os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY', 'your-key-if-not-using-env')
```

```python
folders = glob.glob("knowledge-base/*")
folders
```

```python
# Read in documents using LangChain's loaders
# Take everything in all the sub-folders of our knowledgebase

folders = glob.glob("knowledge-base/*")

# With thanks to CG and Jon R, students on the course, for this fix needed for some users 
text_loader_kwargs = {'encoding': 'utf-8'}
# If that doesn't work, some Windows users might need to uncomment the next line instead
# text_loader_kwargs={'autodetect_encoding': True}

documents = []
for folder in folders:
    doc_type = os.path.basename(folder)
    print(f"doc type: {doc_type}")
    loader = DirectoryLoader(folder, glob="**/*.md", loader_cls=TextLoader, loader_kwargs=text_loader_kwargs)
    folder_docs = loader.load()
    for doc in folder_docs:
        doc.metadata["doc_type"] = doc_type
        documents.append(doc)
```

```python
# chunk overlap is important to avoid cutting straight into a sentence, where both parts would be important together.
text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = text_splitter.split_documents(documents)
```
